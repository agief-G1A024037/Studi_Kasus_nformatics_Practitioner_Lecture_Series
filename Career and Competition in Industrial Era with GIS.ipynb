A. Deskripsi Masalah & Solusi

Mengikuti pengalaman profesional Widi Primadepa dalam tata kelola aset perkotaan di Selandia Baru, manajemen aset publik modern harus terbebas dari perangkat lunak ilegal dan terintegrasi secara spasial (Geospatial Information System). Studi kasus ini menerapkan analisis spasial berbasis pustaka sumber terbuka (GeoPandas & Shapely) untuk melakukan buffering zona bahaya sungai, mengidentifikasi aset publik (jembatan, sekolah, RS) yang masuk zona merah, dan mengekspor hasilnya ke format terbuka GeoJSON untuk peta web (WebGIS).

```python
import geopandas as gpd
from shapely.geometry import Point, LineString

assets_data = {
    'Asset_Name': ['RSUD_Kota', 'Jembatan_Pusat', 'SDN_01', 'Kantor_Pos'],
    'Asset_Type': ['Kesehatan', 'Transportasi', 'Pendidikan', 'Layanan'],
    'geometry': [Point(102.265, -3.795), Point(102.270, -3.790), Point(102.280, -3.810), Point(102.260, -3.800)]
}
gdf_assets = gpd.GeoDataFrame(assets_data, crs="EPSG:4326")

river_line = LineString([(102.260, -3.785), (102.270, -3.795), (102.275, -3.805)])
gdf_river = gpd.GeoDataFrame({'Name': ['Sungai_Utama']}, geometry=[river_line], crs="EPSG:4326")

gdf_assets_m = gdf_assets.to_crs(epsg=32748)
gdf_river_m = gdf_river.to_crs(epsg=32748)

flood_zone_m = gdf_river_m.buffer(500)
gdf_flood_zone = gpd.GeoDataFrame(geometry=flood_zone_m, crs="EPSG:32748")

at_risk_assets = gpd.sjoin(gdf_assets_m, gdf_flood_zone, how="inner", predicate="intersects")

print("=== DAFTAR ASET PUBLIK DALAM ZONA BAHAYA BANJIR (RADIUS 500M) ===")
for _, row in at_risk_assets.iterrows():
    print(f"- {row['Asset_Name']} (Kategori: {row['Asset_Type']}) [STATUS: WASPADA TINGGI]")

gdf_assets.to_file("aset_publik_bengkulu.geojson", driver="GeoJSON")
print("\nData spasial berhasil diekspor ke format 'aset_publik_bengkulu.geojson'!")
```
