A. Deskripsi Masalah & Solusi

Pemerintah daerah menerima ribuan laporan aduan masyarakat terkait infrastruktur publik (jalan berlubang, lampu jalan mati, banjir, tumpukan sampah). Sesuai materi Dr. Andria Arisal, data mentah harus diolah untuk menghasilkan insight pengambilan keputusan. Studi kasus ini menerapkan analisis keranjang belanja (Market Basket Analysis) dengan algoritma Apriori untuk menemukan pola kejadian yang sering muncul secara bersamaan, sehingga dinas terkait dapat melakukan perbaikan pencegahan (preventive maintenance) secara terpadu.

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

laporan_warga = [
    ['Jalan_Berlubang', 'Lampu_Mati', 'Drainase_Mampet'],
    ['Jalan_Berlubang', 'Drainase_Mampet', 'Banjir_Genangan'],
    ['Lampu_Mati', 'Rambu_Rusak'],
    ['Jalan_Berlubang', 'Drainase_Mampet', 'Banjir_Genangan', 'Lampu_Mati'],
    ['Drainase_Mampet', 'Banjir_Genangan'],
    ['Jalan_Berlubang', 'Lampu_Mati']
]

te = TransactionEncoder()
te_ary = te.fit(laporan_warga).transform(laporan_warga)
df = pd.DataFrame(te_ary, columns=te.columns_)

frequent_itemsets = apriori(df, min_support=0.4, use_colnames=True)

rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7)

print("=== TEMUAN POLA ASOSIASI KELUHAN WARGA ===")
for _, row in rules.iterrows():
    antecedents = list(row['antecedents'])
    consequents = list(row['consequents'])
    print(f"Jika ada keluhan: {antecedents} -> Kemungkinan muncul: {consequents}")
    print(f"Support: {row['support']:.2f} | Confidence: {row['confidence']:.2f} | Lift: {row['lift']:.2f}\n")
```
